```python
import pandas as pd
import re
from openpyxl import load_workbook

# 🔥 CLASSIFICATION FUNCTION
def classify(msg):

    msg = msg.lower()

    # DELIVERY
    if "late" in msg or "delay" in msg:
        return "Delivery", "Late Delivery"

    elif "not delivered" in msg or "not received" in msg:
        return "Delivery", "Order Not Delivered"

    # RIDER
    elif "wrong rider" in msg:
        return "Rider", "Wrong Assignment"

    elif "pickup" in msg and "delivery" in msg:
        return "Rider", "Multiple Orders Issue"

    elif "rider" in msg and "rude" in msg:
        return "Rider", "Rider Behavior"

    # FOOD
    elif "cold" in msg:
        return "Food", "Cold Food"

    elif "bad" in msg or "taste" in msg:
        return "Food", "Bad Taste"

    elif "missing" in msg:
        return "Food", "Missing Item"

    elif "wrong item" in msg:
        return "Food", "Wrong Item"

    # HYGIENE
    elif "dirty" in msg or "hair" in msg:
        return "Hygiene", "Contamination"

    # PAYMENT
    elif "refund" in msg:
        return "Payment", "Refund Issue"

    # POSITIVE
    elif "good" in msg or "excellent" in msg:
        return "Feedback", "Positive"

    # DEFAULT
    else:
        return "Customer", "General Complaint"


# 🔥 INPUT SYSTEM
print("05:53 19-03-2026 Rahat Main Counter Rider pickup 2 delivery order at the same time.:")

text = ""
while True:
    line = input()
    if line == "":
        break
    text += line + "\n"

data = []

# 🔥 SMART PARSER (handles messy WhatsApp text)
pattern = r"(\d{1,2}:\d{2})\s*(?:am|pm)?\s*(\d{2}-\d{2}-\d{4})\s*(.*?)\s*(.+)"

matches = re.findall(pattern, text, re.IGNORECASE | re.DOTALL)

for match in matches:
    try:
        time, date, branch, complaint = match

        category, classification = classify(complaint)

        row = {
            "Time": time,
            "Date": date,
            "Branch": branch.strip(),
            "Category": category,
            "Classification": classification,
            "Stations": branch.strip(),
            "Complain": complaint.strip(),
            "Response (Y / N)": "",
            "Manager,s Reply": "",
            "Shift Manager": "",
            "Employe Name": "",
            "Employe ID": "",
            "Fine": "",
            "evidence1": "",
            "evidence2": "",
            "evidence22": "",
            "reply evidence": "",
            "reply evidence2": "",
            "reply back": ""
        }

        data.append(row)

    except:
        continue


df = pd.DataFrame(data)

if df.empty:
    print("❌ No data detected! Check your format.")
    exit()

# 🔥 EXACT COLUMN ORDER
columns = [
    "Time","Date","Branch","Category","Classification","Stations","Complain",
    "Response (Y / N)","Manager,s Reply","Shift Manager",
    "Employe Name","Employe ID","Fine",
    "evidence1","evidence2","evidence22",
    "reply evidence","reply evidence2","reply back"
]

df = df[columns]

# 🔥 FILE NAME (your Feb sheet)
file_name = "Feb_surveillance.xlsm"

try:
    # Load existing file
    book = load_workbook(file_name)
    writer = pd.ExcelWriter(file_name, engine='openpyxl', mode='a', if_sheet_exists='overlay')
    writer.book = book

    sheet = book.active
    start_row = sheet.max_row

    df.to_excel(writer, index=False, header=False, startrow=start_row)

    writer.close()

    print("✅ Data successfully ADDED to Feb_surveillance.xlsm")

except:
    # اگر file نہ ہو
    df.to_excel(file_name, index=False)
    print("✅ New file created: Feb_surveillance.xlsm")
```
